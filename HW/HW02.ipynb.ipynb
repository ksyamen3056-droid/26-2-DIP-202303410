### Gaussian Noise

```python
n_gauss = np.random.normal(0,32,f.shape)
plt.figure(dpi=128)
plt.imshow(n_gauss, cmap='gray', aspect = 'equal')
plt.axis('off')
plt.title('Gaussian noise')
plt.show()
```

### Salt and Pepper Noise

```python
sp = np.random.randint(0,100,np.shape(f))
f_sp = f.copy()
f_sp[sp<3] = 0
f_sp[sp>96] = 255

plt.figure(dpi=128)
plt.imshow(f_sp,cmap='gray',aspect='equal')
plt.axis('off')
plt.show()
```

```python
plt.hist(f.ravel(),256)
plt.title('histogram of corrupted image(f+n)')
plt.show()

plt.hist(f_sp.ravel(),256)
plt.title('histogram of corrupted image(f+n)')
plt.show()
```

### Mean vs Median Filtering

```python
SS = 7 # kernel size (square kernel)
ss = int((SS-1)/2)
r_median = np.zeros(np.shape(f_sp))
r_mean = np.zeros(np.shape(f_sp))
```

```python
for x in range(ss,np.size(f_sp,0)-ss):
  for y in range(ss,np.size(f_sp,1)-ss):
    Sxy = f_sp[x-ss:x-ss+SS,y-ss:y-ss+SS]
    r_mean[x,y] = np.mean(Sxy)
    r_median[x,y] = np.median(Sxy)
```

```python
plt.figure(figsize=(16,9))
plt.subplot(131)
plt.imshow(f_sp,cmap='gray',aspect='equal')
plt.axis('off')
plt.title('corrupted image')
plt.subplot(132)
plt.imshow(r_mean,cmap='gray',aspect='equal')
plt.axis('off')
plt.title('mean filtered')
plt.subplot(133)
plt.imshow(r_median,cmap='gray',aspect='equal')
plt.axis('off')
plt.title('median filtered')
plt.show()
```

## Adaptive Local Noise Reduction

```python
# Adaptive local noise reduction
f = f_n

std_eta = 16
SS = 5

ss = int((SS-1)/2)
r_amean = np.zeros(np.shape(f))
for x in range(ss,np.size(f,0)-ss):
  for y in range(ss,np.size(f,1)-ss):
    Sxy = f[x-ss:x-ss+SS,y-ss:y-ss+SS]
    r_mean[x,y] = np.mean(Sxy)
    std_Sxy = np.std(Sxy)
    r_amean[x,y]=f[x,y]-((std_eta*std_eta)/(std_Sxy*std_Sxy))*(f[x,y]-np.mean(Sxy))
```

```python
plt.figure(figsize=(16,9))
plt.subplot(131)
plt.imshow(f_n,cmap='gray',aspect='equal',vmin=0,vmax=255)
plt.axis('off')
plt.title('corrupted image')
plt.subplot(132)
plt.imshow(r_mean,cmap='gray',aspect='equal',vmin=0,vmax=255)
plt.axis('off')
plt.title('mean filtered')
plt.subplot(133)
plt.imshow(r_amean,cmap='gray',aspect='equal',vmin=0,vmax=255)
plt.axis('off')
plt.title('adaptive local noise reduction')
plt.show()
```

## Adaptive Median Filtering

```python
# Adaptive Median Filtering
#f = f_sp
f = plt.imread(imgpath+'/circuitboard-saltandpep.tif')
SS_vals = np.array([3, 5, 7])
r_amedian = np.zeros(np.shape(f))
r_median = np.zeros(np.shape(f))
r_case = np.zeros(np.shape(f))

ss_max = int((np.max(SS_vals)-1)/2)
for x in range(ss_max,np.size(f,0)-ss_max):
  for y in range(ss_max,np.size(f,1)-ss_max):
    flag_go = 1
    cnt = 0
    z_xy = f[x,y]
    for SS in SS_vals:
      ss = int((SS-1)/2)
      Sxy = f[x-ss:x-ss+SS,y-ss:y-ss+SS]
      z_med = np.median(Sxy)
      z_min = np.min(Sxy)
      z_max = np.max(Sxy)

      if cnt==0:
        r_median[x,y] = z_med

      if z_min<z_med and z_med<z_max:
        if z_min<z_xy and z_xy<z_max:
          r_amedian[x,y] = z_xy
          flag_go = 0
        else:
          r_amedian[x,y] = z_med
          flag_go = 0
      else:
        if SS == np.max(SS_vals):
          r_amedian[x,y] = z_med
          flag_go = 0

      if flag_go==0:
        r_case[x,y] = cnt
        break
      cnt = cnt+1
```

```python
plt.figure(figsize=(16,9))
plt.subplot(131)
plt.imshow(f,cmap='gray',aspect='equal',vmin=0,vmax=255)
plt.axis('off')
plt.title('corrupted image')
plt.subplot(132)
plt.imshow(r_median,cmap='gray',aspect='equal',vmin=0,vmax=255)
plt.axis('off')
plt.title('median filtered')
plt.subplot(133)
plt.imshow(r_amedian,cmap='gray',aspect='equal',vmin=0,vmax=255)
plt.axis('off')
plt.title('adaptive median filtered')
plt.show()
```

```python
plt.imshow(r_case,cmap='jet',interpolation='nearest')
plt.axis('off')
plt.colorbar()
plt.show()
```

- 복원코드를 직접 작성해보기.

```python

import cv2
import matplotlib.pyplot as plt

# 원본 영상 불러오기
f2 = cv2.imread(
    '/content/drive/MyDrive/의료영상처리실습/DIP4E/dentalXray-pepper-noise.tif',
    cv2.IMREAD_GRAYSCALE
)

# Arithmetic Mean Filter 적용
f2_r = cv2.blur(f2, (3, 3))

# 원본 영상과 복원 영상 비교
plt.figure(dpi=128, figsize=(10, 5))

plt.subplot(121)
plt.imshow(f2, cmap='gray')
plt.title('Original: Pepper Noise')
plt.axis('off')

plt.subplot(122)
plt.imshow(f2_r, cmap='gray')
plt.title('Arithmetic Mean Filter')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python

# 2. Median Filter
f2_r_median = cv2.medianBlur(f2, 3)

# 원본 영상과 결과 비교
plt.figure(dpi=128, figsize=(10, 5))

plt.subplot(121)
plt.imshow(f2, cmap='gray')
plt.title('Original: Pepper Noise')
plt.axis('off')

plt.subplot(122)
plt.imshow(f2_r_median, cmap='gray')
plt.title('Median Filter')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python

import numpy as np
import cv2
import matplotlib.pyplot as plt

# 원본 영상을 실수형으로 변환
f2_float = f2.astype(np.float64)

# 주변 영역의 국소 평균과 국소 분산 계산
local_mean = cv2.blur(f2_float, (7, 7))
local_mean_sq = cv2.blur(f2_float ** 2, (7, 7))
local_var = np.maximum(local_mean_sq - local_mean ** 2, 0)

# 잡음 분산 추정
noise_var = np.mean(local_var)

# Adaptive Local Noise Reduction 적용
f2_r_adaptive = f2_float - (
    noise_var / (local_var + 1e-10)
) * (f2_float - local_mean)

# 픽셀값을 0~255 범위로 제한
f2_r_adaptive = np.clip(
    f2_r_adaptive, 0, 255
).astype(np.uint8)

# 원본과 결과 비교
plt.figure(dpi=128, figsize=(10, 5))

plt.subplot(121)
plt.imshow(f2, cmap='gray')
plt.title('Original: Pepper Noise')
plt.axis('off')

plt.subplot(122)
plt.imshow(f2_r_adaptive, cmap='gray')
plt.title('Adaptive Local Noise Reduction')
plt.axis('off')

plt.tight_layout()
plt.show()
```

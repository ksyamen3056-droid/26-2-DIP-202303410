직접 촬영한 사진을 Python으로 불러오고 이미지의 데이터 타입, 공간해상도, 최소값, 최대값 및 메타데이터를 확인한 후 시각화하였다.

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
```

## 1. 이미지 불러오기

```python
# 직접 촬영한 사진 불러오기
img = Image.open("my_photo.jpg")
img_array = np.array(img)

print("사진 불러오기 완료")
```

## 2. 이미지 정보 확인

```python
print("데이터 타입:", img_array.dtype)
print("공간 해상도 (H, W, C):", img_array.shape)
print("최소값:", img_array.min())
print("최대값:", img_array.max())
print("이미지 모드:", img.mode)
print("이미지 크기 (W, H):", img.size)
print("파일 형식:", img.format)
```

## 3. 그 밖의 메타데이터 확인

```python
# EXIF 메타데이터 확인
exif_data = img.getexif()

if exif_data:
    print("EXIF 메타데이터:")
    for tag_id, value in exif_data.items():
        print(tag_id, ":", value)
else:
    print("EXIF 메타데이터가 없습니다.")

# JPEG 기본 정보
print("\nJPEG 기본 메타데이터 키:", list(img.info.keys()))
print("JFIF 버전:", img.info.get("jfif_version"))
print("JFIF 단위:", img.info.get("jfif_unit"))
print("JFIF 밀도:", img.info.get("jfif_density"))
```

## 4. 이미지 시각화

```python
plt.figure(figsize=(10, 7))
plt.imshow(img)
plt.title("My Favorite Photo")
plt.axis("off")
plt.show()
```

## 결과

사진을 정상적으로 불러왔으며, 데이터 타입과 공간해상도, 픽셀 최소·최대값 및 메타데이터를 확인하였다. 마지막으로 사진을 적절한 크기로 시각화하고 제목을 표시하였다.
